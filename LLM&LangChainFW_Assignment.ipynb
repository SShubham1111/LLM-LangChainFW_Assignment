1: What are Large Language Models (LLMs) and how do they function?
- Large Language Models are advanced AI models trained on massive amounts of text data to understand, generate, summarize, translate, and reason with human language. Examples include OpenAI GPT models and Google Gemini.They are built using Transformer architecture, which uses mechanisms like self-attention to understand relationships between words and context in a sequence. LLMs function through training in two stages: pre training learning language patterns from large datasets and fine-tuning specializing for tasks like chatbots, coding, or domain-specific applications.

2: Discuss the impact of LLMs on traditional software development approaches.
- LLMs have accelerated development through AI-assisted coding tools like GitHub Copilot, reducing coding time and improving productivity.They are shifting software development from rule-based programming toward prompt driven and AI-assisted application development.Traditional workflows like debugging, testing, documentation, and API integration are increasingly automated using LLMs.

3: What are the key advantages and limitations of using LLMs in real-world applications?
- They provide automation, scalability, faster content generation, natural language interaction, and support for tasks like summarization, coding, and customer service.
They can produce hallucinations incorrect but confident outputs, inherit biases from training data, and struggle with factual consistency.

5: Compare and contrast Langchain and LamaIndex. What unique problems does each solve
- LangChain is a framework for building LLM-powered applications involving chains, agents, tools, memory, and workflows.
- LlamaIndex focuses on connecting private or external data sources to LLMs through indexing and retrieval, especially for RAG systems.
- LangChain solves orchestration problems—managing multi-step reasoning, tool calling, and agent workflows; LlamaIndex solves data retrieval problems—making structured/unstructured data searchable for LLMs.

```python
# 7 Integrate Langchain with a third-party API (e.g., weather, news) and show how responses can be generated via LLMs.

!pip install langchain langchain-openai requests -q

import requests
from langchain_openai import OpenAI

llm = OpenAI(
    temperature=0.7,
    openai_api_key="YOUR_API_KEY"
)

city = "Hyderabad"
api_key = "YOUR_WEATHER_API_KEY"

url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_key}&units=metric"

data = requests.get(url).json()

if "main" in data:

    weather_info = f"""
    City: {city}
    Temperature: {data['main']['temp']}°C
    Humidity: {data['main']['humidity']}%
    Condition: {data['weather'][0]['description']}
    """

    prompt = f"Summarize this weather information:\n{weather_info}"

    response = llm.invoke(prompt)

    print(response)

else:
    print("Error fetching weather data:", data)
```

```python
# 8. Create a LamaIndex implementation that indexes a local text file and retrieves answers from it.


!pip install llama-index openai -q

from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from llama_index.llms.openai import OpenAI

documents = SimpleDirectoryReader(input_files=["sample.txt"]).load_data()

index = VectorStoreIndex.from_documents(documents)

query_engine = index.as_query_engine(
    llm=OpenAI(api_key="YOUR_API_KEY")
)

response = query_engine.query(
    "What is the main topic discussed in the text file?"
)

print(response)
```

```python
# 9. Demonstrate combining Langchain with LamaIndex to create a simple document-based Q&A chatbot.

!pip install langchain langchain-openai llama-index openai -q

from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from langchain_openai import OpenAI

documents = SimpleDirectoryReader(input_files=["sample.txt"]).load_data()

index = VectorStoreIndex.from_documents(documents)

query_engine = index.as_query_engine()

llm = OpenAI(
    temperature=0.7,
    openai_api_key="YOUR_API_KEY"
)

user_query = "What does the document say about machine learning?"

retrieved_context = str(query_engine.query(user_query))

prompt = f"""
Use the document context below to answer the user question.

Context:
{retrieved_context}

Question:
{user_query}
"""

response = llm.invoke(prompt)

print(response)
```

10: A legal firm wants to use AI to summarize large volumes of legal
documents and retrieve relevant information quickly. Propose a solution using
Langchain and LamaIndex, and explain how it would work in practice.
- Legal documents such as contracts, case files, judgments, and compliance records would be ingested from PDFs, databases, or document repositories and indexed using LlamaIndex. It would structure and organize unstructured legal text into searchable vector indexes for efficient retrieval.
- When a lawyer asks queries like “Find precedents related to breach of contract” or “Summarize clauses about liability,” LlamaIndex would retrieve the most relevant sections from the indexed legal corpus using semantic search and Retrieval-Augmented Generation (RAG).

- LangChain would manage the application workflow by connecting the LLM with retrieval tools, prompt templates, memory, and reasoning chains. It can generate document summaries, compare legal clauses, answer questions, and even support multi-step agent workflows for complex legal research.

- In practice, a lawyer uploads or connects document repositories, LlamaIndex retrieves relevant context, and LangChain sends that context to an LLM to generate concise summaries or answer questions. For example, instead of manually reviewing 500-page case files, the system could summarize key arguments, surface relevant precedents, and reduce research time significantly while improving productivity.
